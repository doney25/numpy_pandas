2. Create two 3×3 matrices:


One filled with all 5s , One with values from 1 to 9


● Add the two matrices


● Multiply them element-wise


● Then find the transpose of the result


```python
A=np.array([[5,5,5],
           [5,5,5],
           [5,5,5]])
B=np.arange(1,10).reshape(3,3)
print(f"A+B:\n{A+B}\n")
result=A*B
print(f"A*B:\n{result}\n")
matrix_trans=result.T
print(f"Transporse matrix:\n{matrix_trans}")
```

4. Generate a 5×4 array with random values between 0 and 1 (float).


● Find the row with the maximum sum of values.


● Normalize each row using the formula: x-min/max-min

```python
arr=np.random.rand(5,4)
print(arr)

row_sum=np.sum(arr,axis=1)
print(f"\nSum of rows: {row_sum}")

print(f"Row with max sum: {arr[np.argmax(row_sum)]}")

row_min=np.min(arr,axis=1,keepdims=True)
row_max=np.max(arr,axis=1,keepdims=True)
normalize=(arr-row_min)/(row_max-row_min)
print(f"\nNormalized:\n {normalize}")
```

5. arr = np.array([1, 3, 5, 3, 7, 9, 1, 5, 5])


● Find all unique elements


● Find the duplicates (elements that appear more than once)

```python
arr=np.array([1,3,5,3,7,9,1,5,5])
print(f"Unique elements: {np.unique(arr)}")

values,count=np.unique(arr,return_counts=True)
duplicates=values[count>1]
print(f"Duplicate elements: {duplicates}")
```

7. Generate an array of shape (3, 4) with values from 0 to 11.


● Reshape it to (2, 6)


● Then flatten it into a 1D array


● Convert the 1D array back to shape (4, 3)

```python
arr=np.arange(0,12).reshape(3,4)
print(f"Original array:\n {arr}")

print("\nReshaped array")
reshaped_array=arr.reshape(2,6)
print(reshaped_array)

print("\nFlatten 1D array")
flatten_array=reshaped_array.flatten()
print(flatten_array)

print("\n4 X 3 matrix")
print(flatten_array.reshape(4,3))
```

8 . Create an array: a = np.array([[1, 2], [3, 4]])


● Append a new row [5, 6]


● Then append a new column [7, 8, 9]

```python
A = np.array([[1, 2], [3, 4]])
print(A)

A=np.vstack((A,[5,6]))
print("\nAppending new row: ")
print(A)

print("\nAppending new column: ")
A=np.hstack((A,[[7],[8],[9]]))
print(A)
```

9.

a = np.array([1, 2, 3]) <br>
b = np.array([4, 5, 6])<br>
c = np.array([7, 8, 9])<br>
● Join them into a single array<br>
● Reshape the result into a 3×3 matrix<br>
● Extract the diagonal elements


```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
c = np.array([7, 8, 9])

joined_array=np.concatenate((a,b,c))
print("Single Array:")
print(joined_array)

reshaped_array=joined_array.reshape(3,3)
print("\nReshaped Array:")
print(reshaped_array)

print("\nDiagonal elements:")
print(np.diag(reshaped_array))
```
