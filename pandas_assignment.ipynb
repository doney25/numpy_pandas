```python
import pandas as pd
import numpy as np
df=pd.read_csv('/content/titanic_dataset.csv')
```

1. Load the Titanic dataset and display the first 10 rows

```python
df.head(10)
```

2. Find the number of missing values in each column.

```python
df.isnull().sum()
```

3. Find the percentage of passengers who survived, grouped by gender.


```python
df.groupby("Sex")["Survived"].mean()*100
```

4. Find the average, median, and standard deviation of passenger ages.


```python
avg=df["Age"].mean()
print("Average:",avg)
_median=df["Age"].median()
print("Median:",_median)
sd=df["Age"].std()
print("Standard deviation:",sd)
```

5. Count how many passengers were in each class (pclass).

```python
df.Pclass.value_counts()
```

6. Show all passengers who were female, under age 18, and survived.

```python
df[(df["Survived"]==1) & (df["Sex"]=="female") & (df["Age"]<18)]
```

7. Replace missing age values with the median age.

```python
df["Age"]=df["Age"].fillna(_median)
df.head(6)
```

8. Create a column age_group with values:<br>
a. 'Child' for age < 18<br>
b. 'Adult' for age 18–60<br>
c. 'Senior' for age > 60<br>
d. Use: apply() or np.where() or custom function

```python
df["age_group"]=df["Age"].apply(
    lambda x: "Child" if x<18
    else "Adult" if 18<x<60
    else "senior")
df.head(10)
```

9. Display the top 5 passengers who paid the highest fare.


```python
data=df.sort_values(by=['Fare'],ascending=False)
data.head(5)
```

10. Find the average fare paid, grouped by both pclass and sex

```python
df.groupby(['Pclass','Sex'])["Fare"].mean()
```

11. Rename the following columns:<br>
a. 'sex' → 'gender'<br>
b. 'fare' → 'ticket_fare'


```python
df.rename(columns={"Sex":"Gender","Fare":"Ticket_fare"},inplace=True)
df.columns
```

12. Count the number of passengers who survived and didn’t survive, grouped by class.


```python
df.groupby(["Pclass",'Survived']).count()['PassengerId']
```
